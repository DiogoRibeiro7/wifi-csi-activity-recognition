# Real-time Inference
Run live activity recognition from hardware.

```
import asyncio

from wifi_activity_recognition.hardware import create_hardware
from wifi_activity_recognition.models import create_model
from wifi_activity_recognition.preprocessing import normalization


async def stream():
    reader = create_hardware("esp32")
    model = create_model("transformer", input_shape=(1, 64), num_classes=5)
    await reader.connect()
    while True:
        csi = await reader.read_packet()
        if csi is None:
            continue
        norm = normalization.zscore(csi.amplitudes)
        pred = model(norm)
        print("prediction", pred.argmax())


# asyncio.run(stream())  # Uncomment for live inference
```
