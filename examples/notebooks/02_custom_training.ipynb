# Custom Training
Train models on custom CSI data.

```
from pathlib import Path

import numpy as np

from wifi_activity_recognition.datasets import Dataset
from wifi_activity_recognition.models import create_model
from wifi_activity_recognition.training import Trainer

# Load your own CSI arrays
csi = np.load(Path("my_data.npy"))
labels = np.load(Path("my_labels.npy"))
dataset = Dataset.from_files(
    "my_data.npy", "my_labels.npy", train_size=0.8, val_size=0.1
)

model = create_model(
    "resnet", input_shape=dataset.input_shape, num_classes=len(dataset.classes)
)
trainer = Trainer(model, dataset)
trainer.train(epochs=5)
```
