# Getting Started
Basic usage demonstration for WiFi Activity Recognition.

```
from wifi_activity_recognition.datasets import Dataset, generate_synthetic_csi
from wifi_activity_recognition.models import create_model
from wifi_activity_recognition.training import Trainer

# Generate synthetic data
data, labels = generate_synthetic_csi(num_samples=100, num_subcarriers=64)
# Split into dataset object
train = (data[:60], labels[:60])
val = (data[60:80], labels[60:80])
test = (data[80:], labels[80:])
dataset = Dataset(train=train, val=val, test=test)

# Create and train a simple model
model = create_model(
    "cnn2d", input_shape=dataset.input_shape, num_classes=len(dataset.classes)
)
trainer = Trainer(model, dataset)
trainer.train(epochs=1)
```
